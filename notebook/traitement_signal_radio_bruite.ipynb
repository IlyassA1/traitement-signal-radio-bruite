## <u>II. Acquisition d'un signal jusqu'à l'oreille humaine</u>

Le radiotélescope de Nançay, est un instrument scientifique qui permet d’étudier l’Univers. Il capte des ondes radio émises par les étoiles et les galaxies, invisibles à l’œil humain. Grâce à sa grande taille, il peut recevoir des signaux très faibles venant de très loin. Ces signaux sont ensuite analysés par les scientifiques pour mieux comprendre l’espace.



### 1) Introduction

Un signal radioastronomique captée par le radiotélescope décimétrique de Nançay doit traverser plusieurs étapes de transformation avant de devenir audible à l'oreille humaine. L'onde électromagnétique originelle oscille à 1,42 milliard de fois par seconde (1420 MHz), une fréquence complètement imperceptible. La chaîne d'acquisition, ramène progressivement ce signal vers des fréquences audibles (20 Hz à 20 kHz) tout en préservant l'information scientifique.

### 2) Signal cosmique et captation par l'antenne

L'onde électromagnétique provient d'une source lointaine (galaxie, etc…) et frappe l'antenne parabolique du radiotélescope après avoir été réceptionnée par le premier jeu de miroirs pivotant puis « rebondit » pour atteindre le second réflecteur fixe et la parabole.
<p></p>

La relation entre longueur d'onde et fréquence est :

$\lambda=\dfrac{c}{f}=\dfrac{3 \times 10^8}{1.42 \times 10^9}=0.21m=21cm$
$\Leftrightarrow f=\dfrac{c}{\lambda}=\dfrac{3 \times 10^8}{0.21}=1.42 \times 10^9Hz=1.42GHz$

L'antenne centralise les ondes sur deux antennes en les dirigeant dans un « cornet » (une sensible aux ondes entre 1 et 1.8GHz et l’autre entre 1.7 et 3.5 GHz). Le champ électromagnétique excite les électrons du conducteur dans le récepteur, générant une très faible tension oscillant à 1420 MHz (de l'ordre du microvolt, µV).
<p></p>

**<u>Signal</u> : Onde EM (Electromagnétique) $\implies$ Tension FR faible (µV)**

### 3) Amplification sans bruit (LNA)

Le signal reçu est noyé dans le bruit thermique de l'électronique. Le LNA (Low Noise Amplifier) amplifie immédiatement le signal avec un gain de 40-60 dB (×10 000 à ×1 000 000) tout en ajoutant très peu de bruit. C'est le composant le plus critique de toute la chaîne, car une amplification rapide protège le signal des dégradations ultérieures. De plus pour limiter encore plus la dégradation, les premiers amplificateurs sont refroidis par un réfrigérateur à hélium sous la barre des -260 °C.
<p></p>



**<u>Signal</u> : Tension FR amplifiée (environ 1-10 mV), toujours à 1420 MHz**

### 4) Transposition de fréquence (mélangeur)

**<u>Problème</u>** : L'oreille humaine n'entend que jusqu'à 20 kHz. Il est impossible de numériser directement du 1,42 GHz avec du matériel standard.
<p></p>

**<u>Solution</u>** : Multiplier le signal amplifié par un signal généré localement (oscillateur local, OL) pour créer une fréquence intermédiaire (FI) audible. Il va donc y avoir un mélangeur qui va entrer en compte.
<p></p>

Il y a une formule mathématique qui décrit un mélangeur.
Mathématiquement, le produit de deux fréquences génère leur somme ET leur différence :
$cos(2 \pi f_1 t) \times cos(2 \pi f_2 t) = \dfrac{1}{2}[cos(2 \pi (f_1 + f_2 )t) + cos(2 \pi (f_1 - f_2 )t)]$

*<u>Exemple concret:</u>*
* <u>Signal reçu:</u> f_FR = 1420.005 MHz
* <u>Oscillateur local:</u> f_OL = 1420.000 MHz
* <u>Fréquence intermédiaire:</u> f_FI = 1420.005 - 1420.000= 5 kHz (audible)

**<u>Signal:</u> Tension oscillante à 5KHz (basse fréquence, facilement numérisable)**

### 5) Numérisation (Convertisseur Analogique Numérique: CAN)

Le signal analogique, centré autour de quelques kHz, est converti en nombres binaires par un convertisseur analogique numérique. D’après le théorème de Nyquist Shannon, la fréquence d’échantillonnage doit être au moins deux fois plus grande que la fréquence maximale du signal. Comme notre signal utile est d’environ 5 kHz, il faudrait au minimum 10 kHz. Dans la pratique, on choisit une fréquence standard audio (par exemple 44,1 kHz), ce qui garantit une numérisation fidèle et permet de stocker le signal sous forme de fichier WAV.

* <u>Fréquence d'échantillonnage :</u> 44,1 kHz ou 48 kHz (standard audio)
* <u>Résolution</u> : 16 bits par échantillon (65 536 niveaux)
* <u>Format :</u> Fichier WAV non compressé

Pour un signal de 5 kHz, 44.1 kHz est largement suffisant (le minimum serait 10 kHz).
<p></p>

**<u>Signal:</u> Suite de nombres entiers stockée dans un fichier WAV sous forme binaire**

### 6) Restitution acoustique (Convertisseur Numérique Analogique: CNA + Haut-parleur)

Le fichier WAV est relu par l'ordinateur et envoyé à un convertisseur numérique-analogique (CNA), qui retransforme les nombres en tension électrique oscillant à 5 kHz. Cette tension alimente un haut-parleur ou un casque audio, dont la membrane vibre et crée une onde acoustique (variation de pression de l'air).
<p></p>

L'oreille humaine perçoit alors une vibration audible à 5 kHz. Le contenu spectral et temporel du signal original est préservé tout au long de la chaîne (si les paramètres sont bien choisis), permettant une analyse scientifique et une vulgarisation simultanée.
<p></p>

**<u>Signal:</u> Onde mécanique (pression acoustique) $\implies$ Perception auditive**

## <u>III. Acquisition d'un signal fortement bruité</u>

### 1) Introduction

Le signal radioastronomique reçu par le radiotélescope décimétrique de Nançay est très faible et fortement bruité. Malgré la présence de zones de protection radioélectrique, le signal utile reste noyé dans un bruit large bande issu des phénomènes naturels et du bruit thermique des composants électroniques. Le fichier audio fourni a volontairement conservé ce bruit afin de rester fidèle au signal réellement capté par l’instrument.
<p></p>

Dans la chaîne d’acquisition principale, le signal est rendu audible après amplification, transposition de fréquence, filtrage analogique et numérisation. Toutefois, lorsque le niveau de bruit est important, cette chaîne peut être améliorée afin d’augmenter le rapport signal sur bruit et de faciliter l’analyse fréquentielle et l’écoute du signal d’intérêt. C’est dans ce contexte qu’une chaîne d’acquisition alternative peut être proposée.

#### Principe de la chaîne alternative :

Le principe de cette chaîne alternative repose sur un filtrage du signal à deux niveaux complémentaires : un filtrage analogique renforcé avant la numérisation, puis un filtrage numérique après la conversion en données numériques. Ces deux étapes exploitent les avantages respectifs des systèmes analogiques et numériques.

### 2) Filtrage analogique avant numérisation :
Après la transposition de fréquence, le signal se situe dans une bande de fréquences audibles: la FI utile (5kHz). Cependant, il contient toujours des composantes indésirables, notamment du bruit hors bande et large bande. Un filtre analogique d’ordre 2, de type RLC, est alors placé avant le convertisseur analogique-numérique.
<p></p>

Ce filtre permet de conserver uniquement la bande fréquentielle contenant le signal utile et d’atténuer les fréquences parasites tout en tant éliminant tout ce qui dépasse une fréquence seuil (environ 10-50kHz selon la conception). Cette étape est essentielle, car elle limite le bruit dès l’entrée de la chaîne numérique et évite les phénomènes de repliement de spectre (aliasing) lors de l’échantillonnage, conformément au théorème de Nyquist-Shannon où les composantes haute fréquence se mélangent à la bande audio et créent du bruit numérique.


### 3) Filtrage numérique après numérisation :

Après la numérisation, un filtrage numérique est appliqué au signal échantillonné. Contrairement au filtre analogique, le filtre numérique est entièrement paramétrable par logiciel. Il permet d’affiner la bande passante, de renforcer l’atténuation du bruit résiduel et d’adapter le traitement en fonction des observations réalisées sur le spectre du signal.
<p></p>

Ce filtrage améliore significativement la lisibilité du signal, aussi bien lors de l’analyse fréquentielle (FFT, spectrogramme) que lors de l’écoute audio.

### Schéma chaîne acquisition avec filtrage

```mermaid
    D -->|Tension FI| E[Filtre analogique]
```

### Mini-conclusion :

La chaîne d’acquisition alternative combine ainsi un filtrage analogique et un traitement numérique plus fin. Cette approche est représentative des systèmes réels de radioastronomie et de télécommunications, où l’objectif est de préserver l’information dans l’audio tout en réduisant l’influence du bruit.
<p></p>

Elle permet d’obtenir un signal plus exploitable, facilitant l’analyse, l’interprétation et la vulgarisation des « sons de l’espace ».

## <u>IV. Analyse avec Python du signal</u>

### 1) Explication

Nous disposons du fichier du son que nous devons étudier. Celui-ci, à la première écoute, est très bruité et on ne perçoit, à l'oreille humaine aucune variation ou démarcation. C'est pourquoi nous allons utiliser le language python qui va nous permettre de décortiquer, parmi ce bruit des variations marquées dans un spectrogramme par un pic. On utilisera ici la fonction *periodogram* afin de faire la transformée de Fourier (FFT) et la fonction *spectrogram* pour générer un spectrogramme de notre son.

### 2) Code Python

Le code Python ci-dessous, une fois executé, nous donne le spectrogramme du son électromagnétique puis un premier graphe de la transformée de Fourier en plan large puis un plan plus proche sur les pics de variations intéressants au milieu du bruit.


````python
from scipy.signal import periodogram    # Import du module signal de scipy
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
from scipy.signal import spectrogram

fs, data = wavfile.read("signal_recu_transpose.wav")

# Si stéréo, ne prendre qu’un canal
if data.ndim > 1:
    data = data[:, 0]

# Calcul du spectrogramme
f, t, Sxx = spectrogram(data, fs, window='hann', nperseg=1024, noverlap=512)

# Affichage du spectrogramme
plt.figure(figsize=(10, 5))
plt.pcolormesh(t, f, 10 * np.log10(Sxx + 1e-10), shading='gouraud', cmap='inferno')
plt.title("Spectrogramme du signal")
plt.ylabel("Fréquence (Hz)")
plt.xlabel("Temps (s)")
plt.colorbar(label="Intensité (dB)")
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 5))
plt.pcolormesh(t, f, 10 * np.log10(Sxx + 1e-10), shading='gouraud', cmap='inferno')
plt.xlim(0, 4)
plt.ylim(0, 2500)
plt.title("Spectrogramme du signal zoomé")
plt.ylabel("Fréquence (Hz)")
plt.xlabel("Temps (s)")
plt.colorbar(label="Intensité (dB)")
plt.tight_layout()
plt.show()


D = 12                           # Durée du signal en seconde
t = np.arange(0, D, 1/(fs))         # Creation de la base temps avec arange


# Calcul de la transformée de Fourier avec la fonction periodogram

f,Pxx = periodogram(data, fs=(fs))



# Affichage du de la transformée de Fourier du signal-----------------

plt.plot(f,Pxx)                  # Affichage via la fonction plot de Matplotlib
plt.xlabel('Fréquence (Hz)')     # Définition de l'axe des abscisses
plt.ylabel('Module')             # Définition de l'axe des ordonnées
plt.grid()
plt.title ('Figure 2 : Module de la Transformée de Fourier du signal', fontsize=14)
plt.show()

plt.plot(f,Pxx)                  # Affichage via la fonction plot de Matplotlib
plt.xlabel('Fréquence (Hz)')     # Définition de l'axe des abscisses
plt.ylabel('Module')             # Définition de l'axe des ordonnées
plt.xlim(1200,1500)
plt.grid()
plt.title ('Figure 2 (zoomée): Module de la Transformée de Fourier du signal', fontsize=14)
plt.show()    # Affichage des courbes
````

Après execution du code, on remarque que notre spectrogramme est remplie de "bruit" avec des saturations de couleurs pixélisées. Cela est normal car notre son est très bruitée à l'écoute. ## <u> I. Introduction</u>

Suite à notre étude de notre signal de l'espace à l'aide de la Transformée de Fourier, nous avons pu déduire les fréquences importantes composant notre signal avec une composante à 1350 Hz et une autre à 1450 Hz parmis du bruit sur le reste de la bande de 44100 Hz. Ainsi nous pouvons passer à l'étape suivante: le filtrage analogique de notre son.

## <u> II. Le Filtre employé </u>

Ici nous souhaitons conserver les fréquences 1350 Hz et 1450 Hz qui sont assez proches l'une de l'autre et entre la plage de 44100 Hz. C'est pourquoi utiliser un filtre passe-haut ou passe-bas ne servira à rien car ils laisseraient des fréquences où le bruit est présent et risquerai d'éliminer une des composantes importantes de notre signal. C'est pourquoi il est plus intéressant de se tourner vers un passe-bande ou un coupe-bande. Or le coupe-bande supprime des fréquences sur un point donné tandis que le passe-bande ne laisse passer qu'une partie des fréquences suivant une fréquence de coupure donnée. On utilisera donc pour notre filtrage de notre signal un passe-bande de second ordre sous forme d'un schéma électrique RLC pour éliminer le bruit autour et ne conserver que nos deux fréquences: 1350 Hz et 1450 Hz.

## <u> III. Détermination de la fonction de transfert </u>

Notre filtre peut-être modéliser par un circuit RLC: c'est-à-dire un circuit électrique composé d'inductances, de condensateurs et de résistances. On se basera sur le schéma suivant:



Sur ce circuit on a donc un condensateur C, une inductance L, une résistance R, une tension d'entrée $V_E$ et une tension de sortie $V_S$. Et pour ces trois trois composants, on peut exprimer leur impédance:
* Condensateur: $\underline{Z_C}$  $= \dfrac{1}{jC\omega}$
* Inductance: $\underline{Z_L}$  $= jL\omega$
* Résistance: $\underline{Z_R}$  $= R$

Or pour pouvoir trouver la fonction de transfert $T(j\omega)$ du circuit, il faut passer par un pont diviseur de tension entre les impédances du condensateur et de l'inductance en fonction de la résistance. D'où un premier calcul de l'impédance cumulée du condensateur et de l'inductance qui sont en série entre eux:
<p></p>

$\underline{Z_P}=\underline{Z_C}+\underline{Z_L}=jL\omega+\dfrac{1}{jC\omega}=\dfrac{1-LC\omega^2}{jC\omega}$
<p></p>

Et on peut donc utiliser le pont diviseur de tension pour exprimer la fonction de transfert:
<p></p>

$\underline{T}(j\omega)=\dfrac{V_S}{V_E}=\dfrac{\underline{Z_R}}{\underline{Z_R}+\underline{Z_P}}=\dfrac{R}{R+\dfrac{1-LC\omega^2}{jC\omega}}=\dfrac{jRC\omega}{1-LC\omega^2+jRC\omega}$ (et on remarque que cette fonction est de second ordre)
<p></p>

## <u> IV. Comparaison avec la forme canonique </u>

Nous avons déduit la fonction de transfert du circuit en utilisant les impédances de chaque composant. Mais il faut maintenant comparer celle-ci par rapport à sa forme canonique. Et la forme canonique d'une fonction de transfert d'un passe-bande en RLC est:
<p></p>

$\underline{T}(j\omega)=\dfrac{j\dfrac{\omega}{Q\omega_0}\times G}{1-\dfrac{\omega^2}{\omega_0^2}+j\dfrac{\omega}{Q\omega_0}}=\dfrac{jRC\omega}{1-LC\omega^2+jRC\omega}$

<p></p>

Par identification on peut en déduire:
* notre fréquence propre: $\dfrac{1}{\omega_0^2}=LC \Leftrightarrow \omega_0^2=\dfrac{1}{LC} \Leftrightarrow \omega_0=\sqrt{\dfrac{1}{LC}}$
* notre fréquence de coupure: $f_0=\dfrac{\omega_0}{2\pi}$
* notre facteur qualité: $\dfrac{1}{Q\omega_0}=RC \Leftrightarrow Q=\dfrac{1}{RC\omega_0}$
<p></p>

  or $\omega_0=\sqrt{\dfrac{1}{LC}}$ 
  <p></p>
  
  d'où: $Q=\dfrac{\dfrac{1}{RC}}{\sqrt{\dfrac{1}{LC}}} \Leftrightarrow Q=\dfrac{\sqrt{LC}}{RC}$

## <u> V. Calculs valeurs composants </u>

### 1) Bande-passante, fréquence de coupure et propre, facteur de qualité

On sait que nos deux fréquences à garder par notre passe-bande sont 1350 Hz et 1450 Hz. Donc on va utiliser comme fréquence de coupure $f_0= 1400 Hz$ et On peut déterminer notre bande-passante c'est-à-dire la plage de fréquence à laisser passer et celle-ci va être comprise entre 1350 et 1450 Hz d'où: $BP = f_{ch}-f_{cb}=1450-1350=100$.
<p></p>

Or la Band-passante BP est égale à $BP=\dfrac{f_0}{Q}$ ainsi: $Q=\dfrac{f_0}{BP}=\dfrac{1400}{100}=14$
<p></p>

De plus nous avions vu que $f_0=\dfrac{\omega_0}{2\pi}$ donc: $\omega_0=f_0\times 2\pi=1400\times 2\pi\approx 8796.46$

### 2) Résistance, Condensateur, Inductance

Comme vu précédemment, $\omega_0=\sqrt{\dfrac{1}{LC}}=8796.46$ d'où: $LC=\dfrac{1}{\omega_0^2}=\dfrac{1}{8796.46^2}\approx 1.29\times 10^{-8}$
<p></p>

Et $Q=\dfrac{1}{RC\omega_0}=14$ d'où: $RC=\dfrac{\sqrt{LC}}{Q}=\dfrac{\sqrt{1.29\times 10^{-8}}}{14}\approx 8.113\times 10^{-6}$
<p></p>

Or ici pour la détermination de la résistance, nous sommes passés par plusieurs calculs et simulation notamment par le logiciel *Falstad* afin de trouver une résistance permettant de nous donner des composants avec des valeurs proches de composants réels et abordables. Ainsi nous allons utiliser ici une résistance de 120 ohms qui nous permettra d'avoir des valeurs cohérentes avec nos deux pics compris dans le passe-bande.
<p></p>

Ainsi notre condensateur vaut, $RC=8.113\times 10^{-6} \Leftrightarrow C=\dfrac{8.113\times 10^{-6}}{120}\approx 6.76\times 10^{-8} F\approx 68 nF$
<p></p>

Et enfin notre inductance va valoir, $LC=1.29\times 10^{-8} \Leftrightarrow L=\dfrac{1.29\times 10^{-8}}{C}=\dfrac{1.29\times 10^{-8}}{6.76\times 10^{-8}}\approx 0.1908 H \approx 190 mH$
<p></p>

Nous avons donc une résistance **R=120 ohms**, un condensateur **C=68 nF** et une inductance **L=190 mH**.

## <u> VI. Traçage des diagrammes de Bode </u>

Le premier code Python ci-dessous nous permet de tracer le diagramme de Bode sur le gain en dB. Nous le comparons par la suite au diagramme de Bode tracer par le logiciel Falstad en fonction de notre circuit RLC de notre filtre auquel nous avons déterminer la valeur des composants:


````python
#importation des fonctions nécessaires
from scipy import signal
import matplotlib.pyplot as plt

#définition des variables
omega0 = 1400
Q = 14

#définition de la fonction de transfert en la décomposant par son numérateur et son dénominateur dans une liste 
#en fonction des degrès (ax²+bx+c --> [a,b,c])
num = [0,(1/(Q*omega0)),0]
den = [(1/(omega0**2)),(1/(Q*omega0)),1]

#initialisation de la fonction de transfert
sys = signal.TransferFunction(num, den )

#assignation de la fonction de transfert dans bode
w, mag, phase = signal.bode(sys)

#Traçage du premier diagramme de Bode sur le gain
plt.figure(figsize=(10, 5))
plt.grid()
plt.title("Diagramme de Bode : Représentation du gain")
plt.xlabel("fréquence (Hz)")
plt.xlim(10,10000)
plt.ylabel("Gain (dB)")
plt.semilogx(w, mag)
````



Et en comparaison, on remarque que les valeurs sont similaires. Par exemple pour une fréquence de $10^3$ Hz on a environ -20 dB de Gain d'après le graphique du code et de même pour notre schéma Falstad. D'autres points et l'allure de courbe sont similaires. Donc notre courbe Python est bel bien juste. Ainsi la suivante qui représente un diagramme de Bode sur la phase en degré est développé par le programme ci-dessous.

````python
#Traçage du deuxième diagramme de Bode sur la phase
plt.figure(figsize=(10, 5))
plt.title("Diagramme de Bode : Représentation de la phase")
plt.xlabel("fréquence (Hz)")
plt.xlim(10,10000)
plt.ylabel("Phase (degré)")
plt.grid()
plt.semilogx(w, phase)  # Bode phase plot
plt.show()
````

Sur le dernier diagramme, celui-ci correspond bien au diagramme de notre circuit RLC. En effet au point 0 degrès, la courbe atteint bel et bien la valeur de notre fréquence de coupure $f_0$.

Dans cette partie, nous allons étudier notre signal plus en profondeur afin de pouvoir enlever le bruit parasite et le décoder. Pour cela nous passerons par trois grandes étapes: le **filtrage numérique**, la **démodulation** et le **décodage**.

## <u> II. Filtrage *numérique* </u>

Dans la partie précédente nous avions utiliser un filtrage dit **analogique** afin de supprimer le bruit parasite qui s'implémentait en plus dans notre signal. Mais le filtrage numérique a ses limites: il est complexe à mettre en oeuvre, à déterminer les composants nécessaires, à adapter et à multiplier.  C'est pourquoi nous allons utiliser un filtrage numérique qui permet, après numérisation du son analogique, de filtrer sur le son numérique qui n'est certes pas exactement le son originel mais reste fidèle à celui-ci suivant le matériel utilisé et la qualité d'enregistrement (pas de quantification, échantillonnage).
<p></p>

Les filtres numériques seront eux plus précis, facile à régler et à adapter ainsi qu'à multiplier en utilisant seulement des lignes de codes. C'est pourquoi ici, nous utiliserons encore une fois le language *Python* afin de filtrer notre son. Et en se fiant à nos recherches précédentes ainsi qu'aux études de transformatéede Fourier, le filtre le plus approprié est le **passe-bande**.
<p></p>

Mais avant de s'arrêter à l'utilisation d'un seul passe-bande, nous allons utiliser un deuxième filtre qui permettra de supprimer le bruit entre nos deux fréquences étudiées. Et pour cela, nous utiliserons un **coupe-bande**.
<p></p>

Ces deux filtres seront simulés par la fonction ***butterworth*** qui nécessite comme paramètre la durée de notre son qui est de 12 sencondes, nos fréquences de coupures 1350 Hz et 1450 Hz et l'ordre de la fonction et ici on utilisera l'ordre 4 le plus puissant et qualitatif. Ainsi en important les bibliothèques nécessaires on obtient le code suivant:


````python
import numpy as np
from scipy.signal import periodogram
import matplotlib.pyplot as plt
from IPython.display import Audio, display
from scipy.signal import butter, filtfilt
from scipy.io import wavfile

#--- Paramètres signal ---
fs, data = wavfile.read("signal_recu_transpose.wav")


T = 12.0 # durée (s)

t = np.arange(0, T, 1/fs)

rng = np.random.default_rng(0) # reproductible



#--- Conception filtre passe-bande Butterworth ---
ordre = 4

#---- fréquences à conserver premier filtre ----
fd = 1350 #fréquence de coupure (Hz)
ff = 1450

#---- fréquences à conserver second filtre ----
fcd = 1350
fcf = 1450


Wn = (fd / (fs/2), ff/ (fs/2)) # normalisée (Nyquist)
Wnc = (fcd / (fs/2), fcf/ (fs/2))


#---- Application fonction butterworth ----
b, a = butter(ordre, Wn, btype="bandpass")
d, c = butter(ordre, Wnc, btype="bandstop")

#--- Filtrage zéro-phase (pas de déphasage) ---

y = filtfilt(b, a, data.T)
z = filtfilt(d, c, y)



#--- Retour audio ---
print("Audio filtré")
display(Audio(data=z, rate=44100)) #audio filtré
print("Audio original")
display(Audio(data=data.T, rate=44100)) #audio original

#---- Création fichier audio son filtré ----
wavfile.write("signal_recu_transpose_filtered.wav",fs, np.int16(z))


#---- Affichage spectre sonore du son filtré et non-filtré ----
win = int(0.050 * fs) # 50 ms

plt.figure(figsize=(9, 3.5))

plt.plot(t[:win], data[:win], label="Entrée x[n]", alpha=0.7)

plt.plot(t[:win], z[:win], label="Sortie y[n] (filtrée)", linewidth=2)



plt.title("Filtrage passe-bande et coupe-bande (Butterworth, ordre 4) — Vue 50 ms")

plt.xlabel("Temps (s)")

plt.ylabel("Amplitude")

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.show()



f,Pxx = periodogram(z, fs=(fs))



#---- Affichage du de la transformée de Fourier du signal filtrée -----

plt.plot(f,Pxx)                  # Affichage via la fonction plot de Matplotlib
plt.xlabel('Fréquence (Hz)')     # Définition de l'axe des abscisses
plt.ylabel('Module')             # Définition de l'axe des ordonnées
plt.grid()
plt.title ('Figure 2 : Module de la Transformée de Fourier du signal filtré', fontsize=14)
plt.show()
plt.xlim(1300, 1500)
plt.plot(f,Pxx)                  # Affichage via la fonction plot de Matplotlib
plt.xlabel('Fréquence (Hz)')     # Définition de l'axe des abscisses
plt.ylabel('Module')             # Définition de l'axe des ordonnées
plt.grid()
plt.title ('Figure 2 (zoomée): Module de la Transformée de Fourier du signal filtré', fontsize=14)
plt.show()    # Affichage des courbes
````

Grâce à ce code nous avons donc pu précisemment et efficacement éliminer le bruit à à l'intérieur et à l'extérieur de nos fréquences de coupures. On remarque aussi sur graphique de l'amplitude en fonction du temps de notre son qu'on obtient un son beaucoup plus exploitable en orange sous forme sinusoïdale comparé à l'original en bleu. Enfin, sur le rendu audio, le son désagréable du bruit a disparu laissant place à nos deux fréquences maîtresses s'alternant en continu.

Cette écriture sera du binaire donc notre obtejctif est de passer de notre courbe actuel à une courbe variant de 0 à 1 comme celle-ci:



Or la modulation/démodulation n'est pas unique: il existe plusieurs types de modulations suivant nos besoins. Nous aurons donc besoin d'analyser notre  courbe ci-dessous afin d'en déduire le type de modulation employée par le messager.

Il existe trois types de modulations: **ASK** (Amplitude-Shiftt-Keying) qui est une modulation jouant sur l' amplitude du signal en générant une porteuse (un signal sinusoïdale pur et continu) et donnant son amplitude en fonction du bit (0 ou 1), **FSK** (Frequency-Shift-Keying) qui est une modulation jouant sur la fréquence du signal en générant deux porteuses avec deux fréquences différentes que prendront chacun des bits, **PSK** (Phase-Shift-Keying) qui elle joue sur la phase du signal en produisant des décalages de phase suivant le bit.
<p></p>

Et au premier coup d'oeil, on pourrait penser au ASK car notre signal a une amplitude varissante. Or celle-ci est irrégulière du début jusqu'à la fin: L'ASK joue sur deux amplitudes pour les bits 0 et 1. Tandis que pour le PSK, on ne voit aucun décalage de phase dans notre signal. Or en générant de nouveau notre spectogramme on peux remarquer nos deux traits de nos deux fréquences 1350 Hz et 1450 Hz en jaune illuminé:

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
from scipy.signal import spectrogram

fs, data = wavfile.read("signal_recu_transpose_filtered.wav")

# Si stéréo, ne prendre qu’un canal
if data.ndim > 1:
    data = data[:, 0]

# Calcul du spectrogramme
f, t, Sxx = spectrogram(data, fs, window='hann', nperseg=1024, noverlap=512)

# Affichage du spectrogramme
plt.figure(figsize=(10, 5))
plt.pcolormesh(t, f, 10 * np.log10(Sxx + 1e-10), shading='gouraud', cmap='inferno')
plt.title("Spectrogramme du signal")
plt.ylabel("Fréquence (Hz)")
plt.xlabel("Temps (s)")
plt.colorbar(label="Intensité (dB)")
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 3))
plt.pcolormesh(t, f, 10 * np.log10(Sxx + 1e-10), shading='gouraud', cmap='inferno')
plt.xlim(0, 4)
plt.ylim(2500, 5000)
plt.title("Spectrogramme du signal zoomé")
plt.ylabel("Fréquence (Hz)")
plt.xlabel("Temps (s)")
plt.colorbar(label="Intensité (dB)")
plt.tight_layout()
plt.show()
````

Et nous pouvons remarquer que ces traits sont reproduits plus hauts et plus visibles d'où le zoom sur ces fréquences (à cause de fréquences perturbatrices). Nous voyons que ceux-ci alternent entre l'un et l'autre sur la recopie du signal. De plus ces alternances sont précises sur seulement les deux fréquences ressemeblant à un message binaire alternant entre les 0 et 1.
<p></p>

C'est pour cela qu'au vu d'une variation de fréquence, nous utiliserons une **démodulation FSK** qui générera deux porteuses: une première comportant la fréquence de **1350 Hz représentant le bit 0** et l'autre de **1450 Hz le bit 1**. Mais les fréquences ne sont pas les seuls paramètres de notre modulation FSK. Il faut définir un **débit binaire** le nombre de bits sur 1 seconde du signal, le **nombre de bits** totaux et la **durée** du signal:

* Le débit peut-être déduit par utilisation du spectogramme vu précédemment: on regarde le nombre de bits c'est-à-dire de saut entre les fréquences sur 1 seconde. Et après plusieurs essais on peut déduire un débit de 10 Bits/s.
* La durée connue est de 12 secondes mais plus précisemment de 12,8 secondes (pour éviter des problèmes de matrices dans le code et gagner en précision).
* Le nombre de bits sera déduit en multipliant la durée du signal par le débit. D'où: $Nbits = 12,8 \times 10 = 128 bits$.

Ainsi nous pouvons nous lancer dans l'étape de génération du code ci-dessous:

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile

#---- Chargement audio -----
fs, data = wavfile.read("signal_recu_transpose_filtered.wav")

    
Ns = int(fs/10)        #Débit binaire
Nbits = int(12.8*10)   #Nombre totaux de bits
N = int(Ns*Nbits) 


#--- Génération des vecteurs temps ----
t1 = np.linspace(0.0, Ns/fs, Ns)  #vecteur temps porteuses

#---- Initialisation des porteuses ----
#Amplitude
A1 = 5
A2 = 5

#Fréquences
fp1 = 1350
fp2 = 1450

#Calcul des sinusoïdes des porteuses
P1 = [A1*np.sin(2*np.pi*fp1*tt) for tt in t1]
P2 = [A2*np.sin(2*np.pi*fp2*tt) for tt in t1]

Porteuse1 = np.tile(P1, Nbits)
Porteuse2 = np.tile(P2, Nbits)

#---- Définition des correspondances des bits aux porteuses ----
bit0 = data*Porteuse1
bit1 = data*Porteuse2

#---- Tableaux stockant les résultats de la méthode des trapèzes ----
y1 = []
y2 = []

i=0

#---- Boucles enchaînant la méthode des trapèzes sur les porteuses ----
for i in range(0,N,Ns):
    y1.append(np.trapezoid(bit1[i:i+Ns]))
    y2.append(np.trapezoid(bit0[i:i+Ns]))

message_demodule_FSK = ''

#----- Comparaison des valeurs des trapèzes -----
for ii in range(0, len(y1)):
    if abs(y1[ii]) > abs(y2[ii]):
        message_demodule_FSK = message_demodule_FSK + '1'   #bit 1 si la valeur correspond à la porteuse 1
    if abs(y1[ii]) <= abs(y2[ii]):
        message_demodule_FSK = message_demodule_FSK + '0'   #bit 2 si c'est la porteuse 2


print("notre message démodulé donne: ",message_demodule_FSK)
````

## <u> IV. Décodage </u>

Nous avons donc obtenu notre message démodulé en une suite binaire qui reste indéchiffrable à première vue. Mais en utilisant des techniques de décodage nous pourrons passer de cette suite de 0 et de 1 à un message concis et compréhensible.
<p></p>

Pour cela nous commencerons par un décodage manchester qui est une technique de décodage binaire permettant de passer de 10 à 1 et 0 à 1 notamment dans notre cas où beaucoup de terme se répète dû à un encodage Manchester avant transmission. D'où notre fonction de décodage:

````python
#---- Fonction de vérification de la chaîne binaire ----
def validate_bits(bits: str):
    if not set(bits).issubset({'0','1'}):
        raise ValueError("La chaîne doit ne contenir que '0' et '1'.")
    return bits

#---- Fonction de décodage Manchester ----
def manchester_decode(mbits: str) -> str:
    mbits = validate_bits(mbits)
    if len(mbits) % 2 != 0:
        raise ValueError("Longueur Manchester impaire : bits manquants.")
    out = []
    for i in range(0, len(mbits), 2):
        pair = mbits[i:i+2]
        if pair == '10':
            out.append('1')
        elif pair == '01':
            out.append('0')
        else:
            raise ValueError(f"Paire Manchester invalide: '{pair}' à l'index {i}")
    return ''.join(out)

man = manchester_decode(message_demodule_FSK)
print("Notre message décodé vaut: ",man)
````

Enfin il ne nous reste plus qu'à convertir notre suite binaire en lettre sous caractères ASCII et UTF-8. Ce sont deux types d'encodage permettant de passer d'associer du bianire à nos caractères comme l'alphabet et les chiffres ainsi que de la ponctuations pour l'ASCII et des caractères supplémentaires pour l'UTF-8 comme certains Emoji ou autres émoticônes et caractères d'autres pays (chinois, japonais (hirigana, katanaka,...).
<p></p>

Nous allons donc utiliser une fonction convertissant les bits en bytes codant les caractères pour l'UTF-8 que nous utiliserons pour englober plus de caractères:

````python
#---- Fonction de transposition de bits en bytes -----
def bytes_from_bits(bits: str) -> bytes:
    bits = validate_bits(bits)
    n = (len(bits)//8)*8
    bits = bits[:n]
    return bytes(int(bits[i:i+8], 2) for i in range(0, n, 8))

#---- Fonction de décodage UTF-8 ----
def utf8_bytes(text: bytes) -> str:

    return text.decode("utf-8")

bytes_back = bytes_from_bits(man)
print("La chaîne de bytes est: ",bytes_back)
utf = utf8_bytes(bytes_back)
print("Notre message est: ",utf)
````

On voit que dans la transcription de notre message on a quelque chose d'incompréhensible. Ce sont des caractères de contrôle non représentés par beaucoup de systèmes modernes. Qui donne ceux-ci:

## <u> V. Déchiffrement </u>

Si notre message est chiffré à l'aide d'une clé XOR, il nous aurait fallu un **crib** c'est-à-dire un terme récurrent apparaissant dans plusieurs signaux avec lequel nous pourrions déterminer la clé XOR. Or ici nous nous ne connaissons rien à propos de notre signal. C'est pourquoi si celui-ci est chiffré nous pouvons tenté de décrypté en utilisons une technique de dite de **BRUTE-FORCING**. Cette technique va venir tenter toutes les possibilités de clé XOR sur notre message chiffré. Si une des clé est la bonne nous le verrons par déduction par un texte lisible ou une suite de chiffre ayant un sens.
<p></p>

Et pour cela nous allons donc utiliser une fonction classique de chiffrage XOR en fonction de clé 8 bits. Cela nous fait donc 256 possiblités de clés, qui seront ici sous forme hexdécimal.

````python
#Fonction de chiffrement XOR 8bits
def xor_bytes(data: bytes, key: int) -> bytes:

    if not (0 <= key <= 255):
        raise ValueError("La clé doit être un entier 0..255")
    k = int(key) & 0xFF
    return bytes((b ^ k) for b in data)
````

### a) Etablissment du message

Ainsi nous pouvons passer à l'implémentation de notre code Python pour d'abord créer notre chaîne de transmission.
````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import periodogram
import matplotlib.pyplot as plt
from IPython.display import Audio, display
from scipy.signal import butter, filtfilt
from scipy.io import wavfile

#---- établissement du message de départ ----
msg = input("Entrez le message que vous souhaitez envoyer: ")
print('Le message de départ est: ', msg)
````

### b) Pipeline

Après avoir établi votre message, celui-ci passera par une **Pipeline**. Cette pipeline est un enchaînement de codage en ASCII, chiffrement XOR et encodage Manchester. L'ASCII passe par une fonction encodage ASCII simple qui permettra de faire fonctionner notre message avec les prochaines fonctions. Le chiffrement XOR se repose sur une clé 8 bits prédéfini qui sera ici *'4F'* soit en décimal *'79'*. Ensuite conformément au schéma, on passe de caractères ASCII à une séquence binaire. Enfin on passe par un encodage en Manchester ce qui constiturera notre message a envoyé avant d'être modulé.

````python
#---- Pipeline ----
# 1-Encodage ASCII

#Fonction d'encodage d'une chaîne de caractères en 'bytes' ASCII
def ascii_bytes(text: str) -> bytes:
    try:
        return text.encode("ascii")
    except UnicodeEncodeError as e:
        raise ValueError(f"Caractère non-ASCII détecté dans texte '{text}': {e}") from e


msg_ascii = ascii_bytes(msg)
print('Le message sous forme ASCII donne: ', msg_ascii)

# 2-Chiffrement XOR 8bits

#Fonction chiffrage en XOR suivant une clé d'entrée 8 bits (en héxadécimal)
def xor_bytes(data: bytes, key: int) -> bytes:

    if not (0 <= key <= 255):
        raise ValueError("La clé doit être un entier 0..255")
    k = int(key) & 0xFF
    return bytes((b ^ k) for b in data)

#Initialisation clé XOR
key = 0x4F

#Chiffrement suivant notre clé 'key'
msg_chiff = xor_bytes(msg_ascii, key)
print("Le message chiffrée par une clé XOR 8 bits valant ", key, " donne: ", msg_chiff)

# 3-Passage en bytes (texte ASCII) en bits (binaire)

#Fonction de passage ASCII en bits
def bits_from_bytes(data: bytes) -> str:
    return ''.join(f'{b:08b}' for b in data)


msg_bin = bits_from_bytes(msg_chiff)
print("Le message chiffrée en binaire donne: ", msg_bin)

# 4-Codage Manchester

#Fonction nécessaire pour validation d'une entrée si suite binaire ou non
#(essentiel pour éviter une erreur dans le codage Manchester fonctionnant sur du binaire)
def validate_bits(bits: str):
    if not set(bits).issubset({'0','1'}):
        raise ValueError("La chaîne doit ne contenir que '0' et '1'.")
    return bits

#Fonction de codage en Manchester
def manchester_encode(bits: str) -> str:
    bits = validate_bits(bits)
    return ''.join('10' if b=='1' else '01' for b in bits)


msg_cod = manchester_encode(msg_bin)
print('Le message codé en Manchester donne: ', msg_cod)
````

Avant de passer à la modulation on passe par une étape intermédiaire qui est la conversion de notre message en binaire qui est stocké dans une variable type de caractères dite 'str' ou 'string'. Nous la convertissons en une liste binaire afin que notre code pour la modulation fonctionne sans accroc et puisse sélectionner les termes un par un.

````python
#Passage en liste (pour la suite du code)
msg_cod_list = []
for i in range(0, len(msg_cod)):
    if msg_cod[i] == '1':
        msg_cod_list.append(1)
    else:
        msg_cod_list.append(0)

print('Notre message codé en Manchester sous une liste donne: ',msg_cod_list)
````

### c) Modulation FSK

Pour la modulation, nous avons choisi la même employée par le son envoyé précédemment: une modulation FSK. Pour une bonne réception et une bonne capture audio par nos ordinateurs, nous avons choisi d'utiliser une vitesse bianire non de 10 baud mais de 5 baud (soit 10 bits/secondes). La fréquence d'échantillonnage Fe est de 44100 Hz valeur universelle. Quant à nos porteuses, nous avons choisi de réutiliser les mêmes fréquences employées précédemment afin de montrer comment celui-ci a été envoyé. Enfin notre audio est chargé à la fin de notre programme, nous permettant de le jouer pour le transmettre par les enceintes de notre ordinateur.

````python
#---- Modulation FSK ----

# 1-Initialistaion des variables
baud = 5                                    #vitesse binaire (bit/s)
Fe = 44100                                  #fréquence d'échantillonnage (Hz)
Ns = int(Fe/baud)
Nbits = len(msg_cod_list)                   #Nombre de bits de notre message codé
N = int(Ns*Nbits)

msg_duplique = np.repeat(msg_cod_list, Ns)  #Duplication de notre message binaire

# 2-Initialisation vecteurs temps
t1 = np.linspace(0.0, Ns/Fe, Ns)
t2 = np.linspace(0.0, N/Fe, N)

# 3-Génération de nos porteuses:

#Paramètres de nos porteuses
A1 = A2 = 1     #Amplitude de nos porteuses
Fp1 = 1350      #Fréquence première porteuse (Hz)
Fp2 = 1450      #Fréquence deuxième porteuse (Hz)


#Génération porteuses
P1 = [A1*np.sin(2*np.pi*Fp1*tt) for tt in t1]
P2 = [A2*np.sin(2*np.pi*Fp2*tt) for tt in t1]

#Lancement de notre modulation FSK

msg_module_FSK = []

for n in range(0, len(msg_cod_list)):
    if (msg_cod_list[n] == 1):
        msg_module_FSK.extend(P1)
    else:
        msg_module_FSK.extend(P2)


# 4-Affichage Graphes

#Graphes de notre message binaire
plt.figure (figsize = (10, 6))
plt.plot(t2, msg_duplique)
plt.xlabel('Temps (s)')
plt.title('Message binaire')
plt.xlim(0, 1)

#Graphes de notre message modulé
plt.figure(figsize = (10, 6))
plt.plot(t2, msg_module_FSK)
plt.xlabel('Temps (s)')
plt.title('Modulation FSK')
plt.xlim(0.15, 0.25)


#---- AFFichage final de notre audio ----
duree = Nbits/baud
print("Le message modulé est de durée: ", duree, "secondes et donne: ")
display(Audio(data=msg_module_FSK, rate=44100))
````

Une fois la librairie installée sur l'ordinateur qui agira comme récpteur, nous pouvons éxecuter le code ci-dessous nous permettant de capturer l'audio par le microphone de notre ordinateur et de le stocker dans un fichier .wav qui pourra être directement réutiliser par le reste du code.

````python
import sounddevice as sd
import numpy as np
import wave
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import periodogram
import matplotlib.pyplot as plt
from IPython.display import Audio, display
from scipy.signal import butter, filtfilt
from scipy.io import wavfile


#Fonction d'enregistrement audio via microphone
def record_audio(filename="message_reçu.wav", duree=duree, frequence=44100):
    try:
        print(f"Enregistrement pendant {duree} secondes...")
        audio_data = sd.rec(int(duree * frequence), samplerate=frequence, channels=1, dtype='int16')
        sd.wait()  # Attendre la fin de l'enregistrement
        print("Enregistrement terminé.")

        # Sauvegarde en WAV
        with wave.open(filename, 'wb') as wf:
            wf.setnchannels(1)  # Mono
            wf.setsampwidth(2)  # 16 bits
            wf.setframerate(frequence)
            wf.writeframes(audio_data.tobytes())

        print(f"Fichier sauvegardé : {filename}")

    except Exception as e:
        print(f"Erreur : {e}")

#Lancement de l'acquisition audio du signal puis écriture sous fichier "message_reçu.wav"
record_audio()
````

### b) Filtrage numérique

Une fois l'enregistrement sauvegardé nous pouvons donc appliquer le schéma inverse de notre transmission en ajoutant un filtrage numérique avant pour s'assurer de n'avoir que les fréquences souhaitées. Mais d'abord on charge le fichier de notre audio nouvellement obtenu puis on le représente sur un graphe pour une meilleure visualisation et remarquer les erreurs possibles lors de l'enregistrement (aucun son, pics involontaires...).

````python
#Chargement de l'audio
fs, data = wavfile.read("message_reçu.wav")

#---- Affichage de l'audio en fonction du temps ----
#Création d'un vecteur temps
t = np.linspace(0, len(data) / fs, len(data))

plt.plot(t, data)
plt.xlabel('Temps (s)')
plt.title('Message reçu')
plt.show()

#------ Filtrage audio ------

#Affichage de la FFT afin de visualiser le bruit et notre message
f, Pxx = periodogram(data, fs=fs)

#Version normale
plt.plot(f, Pxx)
plt.title('Transformée de Fourier du signal')
plt.xlabel('Fréquence (Hz)')
plt.ylabel('Module')
plt.grid(True)
plt.show()
#Version centrée sur les pics importants
plt.plot(f, Pxx)
plt.title('Transformée de Fourier du signal')
plt.xlabel('Fréquence (Hz)')
plt.ylabel('Module')
plt.xlim(250, 450)
plt.grid(True)
plt.show()
````

Nous pouvons donc passer, parès être sûr de l'acquisition, au **filtrage numérique**. Ici, nous nous intéressons aux fréquences 1350 Hz et 1450 Hz qui sont situées au milieu des autres fréquences et ont un écart entre elle. C'est pourquoi nous utiliserons à la fois un filtre Butterworth **passe-bande** (pour exclure les fréquences en dehors) et un filtre **coupe-bande** (pour les fréqunces entre). A la fin on dispose d'un rendu sonore pour comparer les résultats ainsi qu'une transformation de Fourier pour mieux apercevoir la nouvelle composition de notre audio.

````python

#----- Filtrage numérique -----

T = durée # durée (s)
t = np.arange(0, T, 1/fs)


#--- Conception filtre passe-bande et coupe-bande Butterworth ---
#Définition de l'ordre des filtres (allant de 1 à 4)
#un filtre d'ordre 4 est le plus efficace et puissant
ordre = 4

#Définition des fréquences de coupures
fd = 1350
ff = 1450


Wn = (fd / (fs/2), ff/ (fs/2)) # normalisée (Nyquist)
Wnc = (fd / (fs/2), ff/ (fs/2))


#Initialisation des points des filtres
b, a = butter(ordre, Wn, btype="bandpass")   #passe-bande
d, c = butter(ordre, Wnc, btype="bandstop")  #coupe-bande


#--- Filtrage zéro-phase (pas de déphasage) ---

y = filtfilt(b, a, data.T)
w = filtfilt(d, c, y)


#--- Affichage audio après et avant ---
print("Audio filtrée: ")
display(Audio(data=w, rate=44100))
print("Audio de départ: ")
display(Audio(data=data.T, rate=44100))


#--- Affichage avec un zoom de 50ms ---
win = int(0.050 * fs) # 50 ms

plt.figure(figsize=(9, 3.5))

plt.plot(t[:win], data[:win], label="Entrée x[n]", alpha=0.7)

plt.plot(t[:win], w[:win], label="Sortie y[n] (filtrée)", linewidth=2)

plt.title("Filtrage passe-bas (Butterworth, ordre 4, fc=? Hz) — Vue 50 ms")

plt.xlabel("Temps (s)")

plt.ylabel("Amplitude")

plt.legend()

plt.grid(True)

plt.tight_layout()

plt.show()


#Génération de la transformation de Fourier
f, Pxx = periodogram(w, fs=fs)

#---- Affichage Transformation de Fourier ----
#Version normale
plt.plot(f, Pxx)
plt.title('Transformée de Fourier du signal')
plt.xlabel('Fréquence (Hz)')
plt.ylabel('Module')
plt.grid(True)
plt.show()
#Version centrée sur les pics importants
plt.plot(f, Pxx)
plt.title('Transformée de Fourier du signal (zoomée)')
plt.xlabel('Fréquence (Hz)')
plt.ylabel('Module')
plt.xlim(1200, 1600)
plt.grid(True)
plt.show()
````

### c) Démodulation FSK

Ensuite nous pouvons passer à la **démodulation FSK** du message. Comme pour la modulation, on prend les mêmes paramètres: 5 baud, génération de deux porteuses de 1350 et 1450 Hz et d'amplitude 1, association du bit 1 à la porteuse de 1350 Hz et le bit 0 à la porteuse de 1450 Hz.

````python

#------- Démodulation FSK ------

# 1-Initialisation des paramètres
baud = 5               #Vitesse binaire (bits/s)
Ns = int(fs/baud)
Nbits = int(9.6*5)      #Nombre de bits
N = int(Ns*Nbits)
data = w


# 2-Génération des vecteurs temps
t1 = np.linspace(0.0, Ns/fs, Ns)
t2 = np.linspace(0.0, N/fs, N)

# 3-Génération des porteuses

#Paramètres des porteuses
A1 = 1
A2 = 1
fp1 = 1350
fp2 = 1450

#Génération
P1 = [A1*np.sin(2*np.pi*fp1*tt) for tt in t1]
P2 = [A2*np.sin(2*np.pi*fp2*tt) for tt in t1]

#Duplication des porteuses (même effet que np.repeat)
Porteuse1 = np.tile(P1, Nbits)
Porteuse2 = np.tile(P2, Nbits)

#Association des porteuses à un bit respectif
bit1 = data*Porteuse1
bit0 = data*Porteuse2


y1 = []
y2 = []

i=0

for i in range(0,N,Ns):
    y1.append(np.trapezoid(bit1[i:i+Ns]))
    y2.append(np.trapezoid(bit0[i:i+Ns]))

msg_demodule_FSK = ''

for ii in range(0, len(y1)):
    if abs(y1[ii]) > abs(y2[ii]):
        msg_demodule_FSK = msg_demodule_FSK + '1'
    if abs(y1[ii]) <= abs(y2[ii]):
        msg_demodule_FSK = msg_demodule_FSK + '0'


print("Le message démodulé donne: ", msg_demodule_FSK)
````

### d) Pipeline

La dernière étape pour enfin pouvoir lire notre message n'est autre que notre **Pipeline**: décodage **Manchester**, conversion **binaire en ASCII** et déchiffrement **XOR**.

````python

#------ Pipeline -------

# 1-Décodage Manchester

#Fonction de validation du message en binaire
#(essentiel pour éviter des bugs sur le Manchester)
def validate_bits(bits: str):
    if not set(bits).issubset({'0','1'}):
        raise ValueError("La chaîne doit ne contenir que '0' et '1'.")
    return bits

#Fonction de décodage Manchester
def manchester_decode(mbits: str) -> str:
    mbits = validate_bits(mbits)
    if len(mbits) % 2 != 0:
        raise ValueError("Longueur Manchester impaire : bits manquants.")
    out = []
    for i in range(0, len(mbits), 2):
        pair = mbits[i:i+2]
        if pair == '10':
            out.append('1')
        elif pair == '01':
            out.append('0')
        else:
            raise ValueError(f"Paire Manchester invalide: '{pair}' à l'index {i}")
    return ''.join(out)

msg_decod = manchester_decode(msg_demodule_FSK)
print("Le message décodé en Manchester donne: ", msg_decod)

# 2-Transformation en bytes

#Fonction de transposition d'un message binaire en bytes (caractères ASCII)
def bytes_from_bits(bits: str) -> bytes:
    bits = validate_bits(bits)
    n = (len(bits)//8)*8
    bits = bits[:n]
    return bytes(int(bits[i:i+8], 2) for i in range(0, n, 8))

#Décodage
msg_bytes = bytes_from_bits(msg_decod)
print("Le message décodé en ASCII donne: ", msg_bytes)

# 3-Déchiffrement XOR 8bits

#Fonction de chiffrement XOR 8bits
def xor_bytes(data: bytes, key: int) -> bytes:

    if not (0 <= key <= 255):
        raise ValueError("La clé doit être un entier 0..255")
    k = int(key) & 0xFF
    return bytes((b ^ k) for b in data)
# Clé de chiffrment XOR
key = 0x4F

#Déchiffrment
msg_dech = xor_bytes(msg_bytes, key)
#Affichage final
print("Le message original envoyé est: ", msg_dech)
````
